```python
import pandas as pd
import numpy as np
import sklearn

print("pandas:", pd.__version__)
print("numpy:", np.__version__)
print("scikit-learn:", sklearn.__version__)
```

```python
# Change this path if your file has a different name or location
DATA_PATH = "data/WA_Fn-UseC_-Telco-Customer-Churn.csv"

df = pd.read_csv(DATA_PATH)
print("Rows, columns:", df.shape)
```

```python
df.head()
```

```python
# Column names and data types
df.info()
```

```python
# TotalCharges is read as text because a few rows are blank (mostly brand-new customers).
# Convert it to numbers; blanks become NaN so we can see how many there are.
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

print("Missing values per column (only columns that have any):")
missing = df.isna().sum()
print(missing[missing > 0])
```

```python
# How many customers churned? (the target column)
counts = df["Churn"].value_counts()
print(counts)
print()
print((df["Churn"].value_counts(normalize=True) * 100).round(1).astype(str) + " %")
```

```python
# Quick sanity check that the data is ready for the next phase
assert df.shape[0] > 1000, "Dataset should have more than 1,000 rows"
assert "Churn" in df.columns, "Target column Churn is missing"
print("Baseline data load OK:", df.shape[0], "rows,", df.shape[1], "columns")
```
